```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

LOG_FILE = '../../src/results/parking_log.csv'

if os.path.exists(LOG_FILE):
    df = pd.read_csv(LOG_FILE, names=['Timestamp', 'Frame', 'Plate', 'Confidence', 'Status'])
    
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    
    sns.histplot(df['Confidence'], bins=20, kde=True, color='teal', ax=axes[0])
    axes[0].set_title('AI Reading Confidence Distribution')
    axes[0].set_xlabel('Confidence Score')
    
    status_counts = df['Status'].value_counts()
    status_counts.plot(kind='bar', color=['#4CAF50', '#F44336'], ax=axes[1])
    axes[1].set_title('Security Checkpoint Log')
    axes[1].set_ylabel('Number of Vehicles')
    plt.xticks(rotation=0)
    
    plt.tight_layout()
    plt.show()
else:
    print("⚠️ No live log found yet. Run your live_anpr_system.py first to generate data!")
```
